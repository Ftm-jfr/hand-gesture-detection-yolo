```python
yaml_content = """
path: /kaggle/input/6class-finger

train: dataset/images/train
val: dataset/images/val


names:
  0: one_finger
  1: two_fingers
  2: three_fingers
  3: four_fingers
  4: five_fingers
  5: like
"""

with open("/kaggle/working/data.yaml", "w") as f:
    f.write(yaml_content.strip())

print("data.yaml done!")
```

```python
from ultralytics import YOLO

model = YOLO("yolo11n.pt")  

model.train(
    data="/kaggle/working/data.yaml",
    epochs=60,      
    imgsz=640,
    batch=32,
    degrees=90.0,    
    scale=0.5,        
    fliplr=0.5,       
    mosaic=1.0,       
    mixup=0.5 ,        
    name="hand_gesture_6class",
    patience=0,     
    exist_ok=True
)

print("train is done")
```
